Part F – Calculus: Differentiation

```python
def F(x):
    return 40 + 5*x - 0.2*x**2


def dF(x):
    return 5 - 0.4*x


for x in range(1, 11):
    print("Study Hours:", x,
          "F(x):", F(x),
          "F'(x):", dF(x))
```

Part G – Calculus: Integration

```python
import numpy as np

def L(x):
    return 5*x - 0.2*x**2


x_values = np.linspace(0, 5, 1000)
y_values = L(x_values)

area = np.trapezoid(y_values, x_values)

print("Numerical approximation:")
print(area)
```

Part I – Mini Prediction Model

```python
# Input variables
X = df[["Study_Hours",
        "Attendance",
        "Assignment_Score"]].values

# Output variable
Y = df["Final_Score"].values

# Add column of ones for intercept
X = np.column_stack((np.ones(len(X)), X))

# Calculate beta
beta = np.linalg.inv(X.T @ X) @ X.T @ Y

print("Coefficients:")
print(beta)
```
